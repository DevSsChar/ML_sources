## UMAP (Uniform Manifold Approximation and Projection)

**Purpose:** Non-linear dimensionality reduction for visualization and feature extraction.

**Pros:** High performance, preserves global and local structure.

**Cons:** Sensitive to parameters and can vary between runs.

**Applications:** Data visualization, feature extraction, clustering preprocessing.

**Key hyperparameters:**
- `n_neighbors` (default=15): Controls local neighborhood size.
- `min_dist` (default=0.1): Controls minimum distance between points in the embedding.
- `n_components` (default=2): Output dimensionality.

> **Note:** UMAP requires the `umap-learn` package. Install with `pip install umap-learn` if needed.

```
from umap.umap_ import UMAP

# Example instantiation
umap = UMAP(n_neighbors=15, min_dist=0.1, n_components=2)

# To fit: embedding = umap.fit_transform(X)  # X should be your data matrix
```

## t-SNE (t-Distributed Stochastic Neighbor Embedding)

**Purpose:** Nonlinear dimensionality reduction, excellent for visualizing high-dimensional data.

**Pros:** Good at revealing local structure and clusters in high-dim data.

**Cons:** Computationally expensive and sensitive to hyperparameters; can be prone to overfitting or poor global structure.

**Applications:** Visualization, exploratory data analysis, anomaly detection.

**Key hyperparameters:**
- `n_components` (default=2): Output dimension.
- `perplexity` (default=30): Balances attention between local/global.
- `learning_rate` (default=200): Optimization step size.


```
from sklearn.manifold import TSNE

# Example instantiation
tsne = TSNE(n_components=2, perplexity=30, learning_rate=200)

# To fit: embedding = tsne.fit_transform(X)
```

## PCA (Principal Component Analysis)

**Purpose:** Linear dimensionality reduction using orthogonal projections (principal components).

**Pros:** Easy to interpret, reduces noise, fast.

**Cons:** Linear method — may lose nonlinear structure.

**Applications:** Feature extraction, compression, whitening, exploratory analysis.

**Key hyperparameters:**
- `n_components`: Number of principal components to retain.
- `whiten` (default=False): Whether to whiten components.
- `svd_solver` (default='auto'): Algorithm used to compute components.


```
from sklearn.decomposition import PCA

# Example instantiation
pca = PCA(n_components=2)

# To fit and transform: X_pca = pca.fit_transform(X)
```

## DBSCAN (Density-Based Spatial Clustering of Applications with Noise)

**Purpose:** Density-based clustering which can find arbitrarily shaped clusters and identify outliers.

**Pros:** Does not require specifying number of clusters; finds outliers/noise.

**Cons:** Struggles with varying-density clusters and sensitive to `eps` and `min_samples`.

**Applications:** Anomaly detection, spatial clustering.

**Key hyperparameters:**
- `eps` (default=0.5): Maximum radius of neighborhood.
- `min_samples` (default=5): Minimum number of samples in a neighborhood to form a cluster.


```
from sklearn.cluster import DBSCAN

# Example instantiation
dbscan = DBSCAN(eps=0.5, min_samples=5)

# To fit: labels = dbscan.fit_predict(X)
```

## HDBSCAN (Hierarchical DBSCAN)

**Purpose:** Hierarchical density-based clustering that handles varying-density clusters better than DBSCAN.

**Pros:** Better handling of variable-density clusters, more robust cluster selection.

**Cons:** Can be slower and requires the `hdbscan` package.

**Applications:** Large datasets and complex clustering tasks.

**Key hyperparameters:**
- `min_cluster_size` (default=5): Minimum cluster size.
- `min_samples` (default varies): Minimum samples for core points.

> **Note:** Install via `pip install hdbscan` if not already available.


```
import hdbscan

# Example instantiation
clusterer = hdbscan.HDBSCAN(min_cluster_size=5)

# To fit: labels = clusterer.fit_predict(X)
```

## K-Means Clustering

**Purpose:** Centroid-based clustering that partitions data into `k` clusters by minimizing within-cluster variance.

**Pros:** Efficient and simple to implement.

**Cons:** Sensitive to initialization and to the chosen number of clusters `k`.

**Applications:** Customer segmentation, pattern recognition.

**Key hyperparameters:**
- `n_clusters` (default=8): Number of clusters to form.
- `init`: Initialization method (`'k-means++'` or `'random'`).
- `n_init` (default=10): Number of re-runs with different centroid seeds.


```
from sklearn.cluster import KMeans

# Example instantiation
kmeans = KMeans(n_clusters=3, random_state=42)

# To fit: labels = kmeans.fit_predict(X)
```
