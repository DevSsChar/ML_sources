```python
df.columns
```

```python
final_df=pd.merge(df,df1,on='Country Code',how='left')
```

Save the final dataframe

```python
final_df.head()
```

```python
final_df.Country.value_counts()
```

```python
country_counts = final_df.Country.value_counts()
plt.figure(figsize=(12,6))
sns.barplot(x=country_counts.index, y=country_counts.values)
plt.xticks(rotation=90)
plt.xlabel("Country")
plt.ylabel("Number of Restaurants")
plt.title("Number of Restaurants per Country")
plt.show()
```

Pie chart -> First has values, then has labels, We choose top 3 here by slicing,

```python
plt.pie(final_df.Country.value_counts().values[:3], labels=final_df.Country.value_counts().index[:3],autopct="%1.2f%%")
```

Here, we combine Aggregate rating,color and text to see the information about ratings. We use group by and apply size on it. Thus we can infer that,

0.0 Not rated White

1.8->2.4 Poor Red

2.5->3.5 Avg Orange

3.5->3.9 Good Yellow

4.0->4.4 Very Good Green

4.4+ Excellent Dark Green

reset_index()->COnvert to df

rename()->to rename something

columns={'0':'Rating count'} wherever column name is 0 replace by this name

```python
ratings=final_df.groupby(['Aggregate rating','Rating color','Rating text']).size().reset_index().rename(columns={0:'Rating Count'})
```

```python
ratings
```

```python
plt.figure(figsize=(12,8))
sns.barplot(x='Aggregate rating',y='Rating Count',data=ratings)
plt.xticks(rotation=90)
```

```python
plt.figure(figsize=(12,8))
sns.barplot(x='Aggregate rating',y='Rating Count',hue='Rating color',data=ratings,palette=['white','red','orange','yellow','green','green'])
plt.xticks(rotation=90)
```

```python
plt.figure(figsize=(12,8))
sns.countplot(x='Rating color',data=ratings,palette=['blue','red','orange','yellow','green','green'])
# this is the count of records of in ratings, not the count of total records.
# Like white has one record in rating, red has 4-5. Just to check bands
```
