```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("elikplim/eergy-efficiency-dataset")

print("Path to dataset files:", path)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df=pd.read_csv('/kaggle/input/eergy-efficiency-dataset/ENB2012_data.csv')
df
```

```python
df.columns=['Relative Compactness','Surface Area','Wall Area','Roof Area','Overall Height','Orientation',
            'Glazing Area','Glazing Area Distribution','Heating Load','Cooling load']
df
```

```python
df.info()
df.describe()
```

```python
df.isna().sum()
print("Nulls:",df.isnull().sum())
```

```python
X=df.drop(['Heating Load','Cooling load'],axis=1)
y=df[['Heating Load', 'Cooling load']]
X
y
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score,accuracy_score

X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)

scaler=StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
```

```python
lr=LinearRegression()

lr.fit(X_train_scaled,y_train)

y_pred=lr.predict(X_test_scaled)
y_pred
```

```python
print("Mean Squared Error:", mean_squared_error(y_test, y_pred))
print("R-squared:", r2_score(y_test, y_pred))
```
