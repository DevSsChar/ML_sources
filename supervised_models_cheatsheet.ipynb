# Common Supervised Learning Models — Cheat Sheet

This notebook documents common supervised learning models and associated utility functions. Each section contains a brief description and a minimal code snippet showing how to instantiate the model or use the function.

## One-vs-One Classifier (using Logistic Regression)


**Process:** Trains one classifier for each pair of classes (pairwise classification).  
**Key hyperparameters:** `estimator` (base classifier, e.g., logistic regression).  
**Pros:** Can work well for small numbers of classes and often yields good performance for some multiclass problems.  
**Cons:** Computationally expensive as number of classes grows (O(k^2) classifiers).  
**Common applications:** Multiclass classification when number of classes is small.


```
from sklearn.multiclass import OneVsOneClassifier
from sklearn.linear_model import LogisticRegression
model = OneVsOneClassifier(LogisticRegression())
model
```

## One-vs-All / One-vs-Rest Classifier (using Logistic Regression)


**Process:** Trains one classifier per class; each classifier distinguishes one class vs all others.  
**Key hyperparameters:** `estimator` (base classifier). For `LogisticRegression`, `multi_class='ovr'` does the same.  
**Pros:** Simpler and more scalable than One-vs-One.  
**Cons:** May be less accurate for highly imbalanced classes.


```
from sklearn.multiclass import OneVsRestClassifier
from sklearn.linear_model import LogisticRegression
model = OneVsRestClassifier(LogisticRegression())
# OR
model_ova = LogisticRegression(multi_class='ovr')
model, model_ova
```

## Decision Tree Classifier


**Process:** Tree-based classifier that splits data based on feature values.  
**Key hyperparameters:** `max_depth`, `criterion`, `min_samples_split`, `min_samples_leaf`.  
**Pros:** Easy to interpret and visualize.  
**Cons:** Prone to overfitting if not pruned/controlled.


```
from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier(max_depth=5)
model
```

## Decision Tree Regressor


**Process:** Tree-based regressor that predicts continuous values by splitting feature space.  
**Key hyperparameters:** `max_depth`, `criterion`, `min_samples_leaf`.  
**Pros:** Handles nonlinear relationships and is easy to interpret.  
**Cons:** Can overfit and be sensitive to noisy data.


```
from sklearn.tree import DecisionTreeRegressor
model = DecisionTreeRegressor(max_depth=5)
model
```

## Linear SVM Classifier


**Process:** Finds a linear hyperplane that maximally separates classes.  
**Key hyperparameters:** `C` (regularization), `kernel` (if using SVC), `class_weight`.  
**Pros:** Effective in high-dimensional spaces.  
**Cons:** Not ideal for nonlinear problems without kernels.


```
from sklearn.svm import SVC
model = SVC(kernel='linear', C=1.0)
model
```

## K-Nearest Neighbors Classifier (KNN)


**Process:** Classifies samples based on majority label among nearest neighbors.  
**Key hyperparameters:** `n_neighbors`, `weights`, `algorithm`.  
**Pros:** Simple and effective for small datasets.  
**Cons:** Computationally expensive for large datasets; sensitive to feature scaling.


```
from sklearn.neighbors import KNeighborsClassifier
model = KNeighborsClassifier(n_neighbors=5, weights='uniform')
model
```

## Random Forest Regressor


**Process:** Ensemble of decision trees (bagging) to reduce variance and improve generalization.  
**Key hyperparameters:** `n_estimators`, `max_depth`, `max_features`.  
**Pros:** Robust and less prone to overfitting than single trees.  
**Cons:** Increased complexity and slower prediction time compared to single trees.


```
from sklearn.ensemble import RandomForestRegressor
model = RandomForestRegressor(n_estimators=100, max_depth=5)
model
```

## XGBoost Regressor


**Process:** Gradient boosting algorithm that builds trees sequentially to correct previous errors.  
**Key hyperparameters:** `n_estimators`, `learning_rate`, `max_depth`.  
**Pros:** High accuracy and widely used in competitions.  
**Cons:** Computationally intensive and requires careful tuning.

**Note:** If `xgboost` is not installed, install it via `pip install xgboost` before running the cell below.


```
# XGBoost example (requires xgboost installed)
import xgboost as xgb
model = xgb.XGBRegressor(n_estimators=100, learning_rate=0.1, max_depth=5)
model
```
